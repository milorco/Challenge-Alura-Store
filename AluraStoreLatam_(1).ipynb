# **Estructura de datos**

## El conjunto de datos incluye la siguiente información:

### **Producto y Categoría:** Artículos vendidos y sus calificaciones.

### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)
```

#1. Análisis de facturación



```python
import matplotlib.pyplot as plt
```

```python
anios = list(map(str, range(2020, 2024)))
anios
```

```python
tiendas = df_tiendas[['Tiendas', 'Fecha de Compra', 'Precio']]
tiendas.set_index('Fecha de Compra', inplace=True)
```

```python
def facturacion(df):
  df = df.groupby('Fecha de Compra')['Precio'].sum()/1000000
  df = pd.DataFrame(df)
  return df
```

```python
fact_tienda = facturacion(tienda)
fact_tienda2 = facturacion(tienda2)
fact_tienda3 = facturacion(tienda3)
fact_tienda4 = facturacion(tienda4)
```

```python
fig, axs = plt.subplots(2,2, figsize=(12,8))
fig.subplots_adjust(hspace=0.5, wspace = 0.3)
fig.suptitle('FACTURACION POR TIENDA \n(2020 - 2023)',
             fontsize= 25, y= 1.05, x= 0.5, weight='bold',
             color= 'teal')

axs[0,0].plot(anios, fact_tienda['Precio'], marker= 'o', color= 'firebrick')
axs[0,0].set_title('TIENDA 1', fontsize= 20, pad= 10,
                   fontdict={'weight': 'bold'})

axs[0,1].plot(anios, fact_tienda2['Precio'], marker= 'o', color= 'forestgreen')
axs[0,1].set_title('TIENDA 2', fontsize= 20, pad= 10,
                   fontdict={'weight': 'bold'})

axs[1,0].plot(anios, fact_tienda3['Precio'], marker= 'o', color= 'forestgreen')
axs[1,0].set_title('TIENDA 3', fontsize= 20, pad= 10,
                   fontdict={'weight': 'bold'})

axs[1,1].plot(anios, fact_tienda4['Precio'], marker= 'o', color= 'firebrick')
axs[1,1].set_title('TIENDA 4', fontsize= 20, pad= 10,
                   fontdict={'weight': 'bold'})

for ax in axs.flat:
  ax.set_xlabel('Año', fontsize= 15)
  ax.set_ylabel('Facturacion ($MM)', fontsize= 15)
  ax.grid()



plt.show()
```

### Viendo las graficas de facturacion por Tienda, se puede observar que las Tiendas 1 y 4 tienen una tendencia negativa por lo que es muy probable que las ventas vayan a seguir bajando en el 2023. Sin embargo, para las tiendas 2 y 3, se ve mas una tendencia a la alza en las ventas ya que van mejorando las ventas a comparacion del 2020.

```python
#Obtener la facturacion Total por tienda
fact_tiendas = df_tiendas.groupby('Tiendas')['Precio'].sum()/1000000
fact_tiendas = pd.DataFrame(fact_tiendas)
fact_tiendas = fact_tiendas.round(2)
fact_tiendas
```

```python
#Graficar la facturacion promedio por tienda para comparar visualmente entre tiendas

colores = ['forestgreen', 'forestgreen', 'firebrick',
           'firebrick', 'brown', 'slateblue','gray', 'olive',
           'navy', 'teal', 'tomato']

fig, ax = plt.subplots(figsize= (12, 5))

ax.bar(fact_tiendas.index, fact_tiendas['Precio'], color = colores)
ax.set_title('Facturacion por Tienda\n(2020 - 2023)', fontsize= 25,
             y= 1.05, x= 0.5, color = 'teal', weight='bold')

ax.set_xlabel('')
ax.set_ylabel('Facturacion ($MM)', fontsize = 14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

for i,j in enumerate(fact_tiendas['Precio']):
  ax.text(i, j+10, str(j), color= 'black', fontsize= 15, ha = 'center')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```

# 2. Ventas por categoría

```python
df_tiendas.head(2)
```

```python
categ_tiendas = df_tiendas[['Tiendas', 'Fecha de Compra', 'Categoría del Producto', 'Precio']]
categ_tiendas.set_index('Fecha de Compra', inplace=True)
```

```python
def categoria(df):
  df = df.groupby('Categoría del Producto')['Precio'].sum()/1000000
  df = pd.DataFrame(df)
  return df
```

```python
categ_tienda = categoria(tienda).sort_values('Precio').round(2)
categ_tienda2 = categoria(tienda2).sort_values('Precio').round(2)
categ_tienda3 = categoria(tienda3).sort_values('Precio').round(2)
categ_tienda4 = categoria(tienda4).sort_values('Precio').round(2)
```

```python
categ_tienda
```

```python
col_categ_tienda = tienda['Categoría del Producto'].unique().tolist()
col_categ_tienda2 = tienda['Categoría del Producto'].unique().tolist()
col_categ_tienda3 = tienda['Categoría del Producto'].unique().tolist()
col_categ_tienda4 = tienda['Categoría del Producto'].unique().tolist()
```

```python
def categ_subplots_barh(df_valor, df_valor2, df_valor3, df_valor4, titulo, unidad_medida):

  #Definir variables para lienzo y grafica(s)
  fig, axs = plt.subplots(2,2, figsize=(20,10))

  #Ajustar lienzo
  fig.subplots_adjust(hspace=0.4, wspace = 0.8)
  fig.suptitle(f'{titulo} POR CATEGORIA DEL PRODUCTO \n(2020 - 2023)',
              fontsize= 30, y= 1.05, x= 0.5, weight='bold',
              color= 'teal')

  #Graficar cada sbplot
  axs[0,0].barh(col_categ_tienda, df_valor, color= 'forestgreen')
  axs[0,0].set_title('TIENDA 1', fontsize= 20, pad= 10,
                    fontdict={'weight': 'bold'})

  axs[0,1].barh(col_categ_tienda2, df_valor2, color= 'darkcyan')
  axs[0,1].set_title('TIENDA 2', fontsize= 20, pad= 10,
                    fontdict={'weight': 'bold'})

  axs[1,0].barh(col_categ_tienda3, df_valor3, color= 'mediumaquamarine')
  axs[1,0].set_title('TIENDA 3', fontsize= 20, pad= 10,
                    fontdict={'weight': 'bold'})

  axs[1,1].barh(col_categ_tienda4, df_valor4, color= 'orange')
  axs[1,1].set_title('TIENDA 4', fontsize= 20, pad= 10,
                    fontdict={'weight': 'bold'})

  #Ajustes de subplots
  for ax in axs.flat:
    ax.set_xlabel(f'{titulo}', fontsize= 15)
    ax.set_ylabel('Categoría del Producto', fontsize= 15, weight='bold')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.set_frame_on(False)
    ax.tick_params(axis='y', which='both', size=0)
    ax.xaxis.set_tick_params(labelsize=12)
    ax.yaxis.set_tick_params(labelsize=12)
    ax.xaxis.set_visible(False)
    ax.grid()

  # Agregar etiquetas a las barras
  for i,j in enumerate(df_valor):
    axs[0,0].text(j+6, i, str(j) + unidad_medida, color= 'black', fontsize= 12, ha='left', va='center')

  for i,j in enumerate(df_valor2):
    axs[0,1].text(j+6, i, str(j) + unidad_medida, color= 'black', fontsize= 12, ha='left', va='center')

  for i,j in enumerate(df_valor3):
    axs[1,0].text(j+6, i, str(j) + unidad_medida, color= 'black', fontsize= 12, ha='left', va='center')

  for i,j in enumerate(df_valor4):
    axs[1,1].text(j+6, i, str(j) + unidad_medida, color= 'black', fontsize= 12, ha='left', va='center')

  #Ajustar Marco
  plt.tight_layout(pad= 1.0, h_pad= 1.5, w_pad= 1.5)
  plt.subplots_adjust()

  #Mostrar graficas
  plt.show()
```

```python
categ_subplots_barh(categ_tienda['Precio'], categ_tienda2['Precio'], categ_tienda3['Precio'], categ_tienda4['Precio'], 'FACTURACION', ' MM')
```

### Se puede observar que las CATEGORIAS DEL PORODUCTO con mayor facturacion son los INSTRUMENTOS MUSICALES, LIBROS y DEPORTES Y DIVERSION. Sin embargo, los ultimos 3 con menos Facturacion son ELECTRONICOS, MUEBLES y JUGUETES.

```python
#Obtener la cantidad de ventas por categoria de Porducto
cant_categ_tienda = tienda['Categoría del Producto'].value_counts().sort_values()
cant_categ_tienda2 = tienda2['Categoría del Producto'].value_counts().sort_values()
cant_categ_tienda3 = tienda3['Categoría del Producto'].value_counts().sort_values()
cant_categ_tienda4 = tienda4['Categoría del Producto'].value_counts().sort_values()
```

```python
cant_categ_tienda
```

```python
#Graficar ventas por Categoria del producto
categ_subplots_barh(cant_categ_tienda, cant_categ_tienda2, cant_categ_tienda3, cant_categ_tienda4, 'VENTA', '')
```

### Se puede observar en las graficas que en todas las Tiendas, el top 3 de CATEGORIAS DEL PRODUCTO mas vendidos son los INSTRUMENTOS MUSICALES, LIBROS y DEPORTES Y DIVERSION. Sin embargo, los ultimos 3 con menos ventas son ELECTRONICOS, MUEBLES y JUGUETES.

# 3. Calificación promedio de la tienda


```python
df_tiendas.head(2)
```

```python
calificacion_promedio_tienda = df_tiendas.groupby('Tiendas')['Calificación'].mean().round(2).sort_values(ascending= False)
calificacion_promedio_tienda = pd.DataFrame(calificacion_promedio_tienda)
calificacion_promedio_tienda
```

```python
import seaborn as sns

# Configurar el estilo de Seaborn
sns.set(style="whitegrid")

# Crear el gráfico de barras
ax = sns.barplot(x='Tiendas', hue='Tiendas', y='Calificación', data=calificacion_promedio_tienda, palette= "hls", legend= False)
ax.set_title('Calificación promedio por Tienda', fontsize= 25, y= 1.05, x= 0.5, color = 'teal', weight='bold')
ax.set_xlabel('Tiendas', fontsize= 15)
ax.set_ylabel('Calificación', fontsize= 15)
ax.grid(False)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Ajustar el rango del eje Y
plt.ylim(0, 4.5)  # Establecer el límite inferior en 0 y el superior en 5

# Mostrar los valores sobre las barras
for container in ax.containers:
    ax.bar_label(container)

plt.show()
```
