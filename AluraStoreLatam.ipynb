### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)
```

```python
#Crear columna nueva para identificar el numero de tienda que es.
tienda['Tiendas'] = 'Tienda1'
tienda2['Tiendas'] = 'Tienda2'
tienda3['Tiendas'] = 'Tienda3'
tienda4['Tiendas'] = 'Tienda4'
```

```python
# Unir los 4 DataFrames para generar uno solo.
df_tiendas = pd.concat([tienda, tienda2, tienda3, tienda4])
```

```python
#Cambiar el tipo de dato de "Fecha de Compra" a fecha.
df_tiendas['Fecha de Compra'] = pd.to_datetime(df_tiendas['Fecha de Compra'])
tienda['Fecha de Compra'] = pd.to_datetime(tienda['Fecha de Compra'], dayfirst=True)
tienda2['Fecha de Compra'] = pd.to_datetime(tienda2['Fecha de Compra'], dayfirst=True)
tienda3['Fecha de Compra'] = pd.to_datetime(tienda3['Fecha de Compra'], dayfirst=True)
tienda4['Fecha de Compra'] = pd.to_datetime(tienda4['Fecha de Compra'], dayfirst=True)
```

```python
# Dar formato para obtener el año y mes
df_tiendas['Fecha de Compra'] = df_tiendas['Fecha de Compra'].dt.strftime('%Y')
tienda2['Fecha de Compra'] = tienda2['Fecha de Compra'].dt.strftime('%Y')
tienda3['Fecha de Compra'] = tienda3['Fecha de Compra'].dt.strftime('%Y')
tienda4['Fecha de Compra'] = tienda4['Fecha de Compra'].dt.strftime('%Y')
tienda['Fecha de Compra'] = tienda['Fecha de Compra'].dt.strftime('%Y')
```

#1. Análisis de facturación



```python
#Facturacion promedio por Tienda
facturacion_prom = df_tiendas.groupby('Tiendas')['Precio'].mean().round(2)
facturacion_prom
```

```python
import matplotlib.pyplot as plt
```

```python
anios = list(map(str, range(2020, 2024)))
anios
```

```python
tiendas = df_tiendas[['Tiendas', 'Fecha de Compra', 'Precio']]
tiendas.set_index('Fecha de Compra', inplace=True)
```

```python
def facturacion(df):
  df = df.groupby('Fecha de Compra')['Precio'].sum()/1000000
  df = pd.DataFrame(df)
  return df
```

```python
fact_tienda = facturacion(tienda)
fact_tienda2 = facturacion(tienda2)
fact_tienda3 = facturacion(tienda3)
fact_tienda4 = facturacion(tienda4)
```

```python
fig, axs = plt.subplots(2,2, figsize=(10,6))
fig.subplots_adjust(hspace=0.5, wspace = 0.3)
fig.suptitle('Facturacion por Tienda')

axs[0,0].plot(anios, fact_tienda['Precio'])
axs[0,0].set_title('Facturacion ($MM) Tienda 1')

axs[0,1].plot(anios, fact_tienda2['Precio'])
axs[0,1].set_title('Facturacion ($MM) Tienda 2')

axs[1,0].plot(anios, fact_tienda3['Precio'])
axs[1,0].set_title('Facturacion ($MM) Tienda 3')

axs[1,1].plot(anios, fact_tienda4['Precio'])
axs[1,1].set_title('Facturacion ($MM) Tienda 4')

for ax in axs.flat:
  ax.set_xlabel('Año')
  ax.set_ylabel('Facturacion ($MM)')
  ax.grid()

plt.show()
```

```python
fact_tiendas = df_tiendas.groupby('Tiendas')['Precio'].sum()/1000000
fact_tiendas = pd.DataFrame(fact_tiendas)
fact_tiendas = fact_tiendas.round(2)
fact_tiendas
```

```python
colores = ['royalblue', 'orange', 'forestgreen', 'red',
           'purple', 'brown', 'slateblue','gray', 'olive',
           'navy', 'teal', 'tomato']

fig, ax = plt.subplots(figsize= (12, 5))

ax.bar(fact_tiendas.index, fact_tiendas['Precio'], color = colores)
ax.set_title('Facturacion por Tienda\n(2020 - 2023)', fontsize= 20)

ax.set_xlabel('')
ax.set_ylabel('Facturacion ($MM)', fontsize = 14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

for i,j in enumerate(fact_tiendas['Precio']):
  ax.text(i, j+10, str(j), color= 'black', fontsize= 15, ha = 'center')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```
